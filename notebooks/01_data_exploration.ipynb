```python
import mne
import matplotlib.pyplot as plt

file_path = "../data/raw/S001R04.edf"

raw = mne.io.read_raw_edf(
    file_path,
    preload=True
)
```

```python
print(raw)
```

```python
print("Number of channels:", len(raw.ch_names))
print("Sampling frequency:", raw.info["sfreq"], "Hz")
print("Duration:", raw.times[-1], "seconds")
print("Number of samples:", raw.n_times)
```

```python
print(raw.ch_names)
```

```python
raw.plot(
    n_channels=10,
    duration=10,
    scalings="auto"
)
```

```python
print(raw.annotations)
```

```python
for onset, duration, description in zip(
    raw.annotations.onset,
    raw.annotations.duration,
    raw.annotations.description
):
    print(
        f"Onset: {onset:.2f}s | "
        f"Duration: {duration:.2f}s | "
        f"Event: {description}"
    )
```

```python
# # Event Exploration
# annotationهای T0/T1/T2 را به eventهای عددی تبدیل کردم تا در مراحل بعد بتوانم EEG را بر اساس زمان وقوع task به epochهای قابل تحلیل تقسیم کنم.
events, event_id = mne.events_from_annotations(raw)

print("Event IDs:")
print(event_id)

print("\nEvents shape:")
print(events.shape)

print("\nFirst 10 events:")
print(events[:10])
```

```python
print("Total events:", len(events))

print("T0 events:", sum(events[:, 2] == event_id["T0"]))
print("T1 events:", sum(events[:, 2] == event_id["T1"]))
print("T2 events:", sum(events[:, 2] == event_id["T2"]))
```

```python
for event in events[:5]:
    sample_index = event[0]
    event_code = event[2]

    print(
        "Sample:", sample_index,
        "| Time:", sample_index / raw.info["sfreq"],
        "| Code:", event_code
    )
```
